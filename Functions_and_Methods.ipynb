```python
# 1️⃣ Defining & Calling a Function
def greet():
    print("Hello, welcome to python class")

greet()
greet()
greet()
```

```python
# Calling Function Inside Loop
for _ in range(3):
    greet()
```

```python
# 2️⃣ Function with Parameters (Inputs)
def welcome(name):
    print(f'Hello {name}, welcome to python class')

welcome('John') # passing arguments
welcome('Jane')
```

```python
# 3️⃣ Multiple Parameters
def welcome(first_name,last_name):
    print(f'Hello {first_name} {last_name}, welcome to python class')

welcome('John', 'Doe')
welcome('Jane', 'Doe')
```

```python
# 4️⃣ Default Parameter
def printer(color = 'Black'):
  print(f'Print Document in {color} color')

printer('Blue');
printer();
printer('Red');
```

```python
# 5️⃣ Keyword Arguments
def book_details(name, author, year):
    print(f"The Book '{name}' is written by author {author} in {year}")

book_details('Harry Potter','JK Rowling',1997)
# book_details("JK Rowling","Harry Potter",1997) # On passing mutiple argument mistakes may happen in order to avoid we can user keyword argument
book_details(author='JK Rowling', name='Harry Potter', year=1997)
```

```python
# 6️⃣ Function with Return Value
def add(a,b):
    return a + b

result = add(5,22)
    
print(result)
```

```python
# 7️⃣ *Variable-Length Arguments → args
# Used when we don’t know how many arguments will be passed.
def total(*numbers):
    total_sum = 0
    for num in numbers:  # we use for loop which we learned in loops concept  
        total_sum +=num # we use Addition Assignment which we learned in Assignment Operators in Operators
    print(total_sum)

# even short way
def short_total(*numbers):
    return sum(numbers)

total(10, 20, 30, 40)
output = short_total(10, 20)
print(output)

# another example
# global keyword lets us modify the outer variable starting_id
# Without global, starting_id inside the function would be treated as a separate local variable
starting_id = 1000 
def student_id_card_generator(*names):
    # starting_id = 1000 # local variable will reset the id
    global starting_id
    for name in names:
        print(f'Student Id {starting_id} for {name}')
        starting_id += 1

student_id_card_generator('Madhu','Aravind','Rajesh','Hemanth')
```

```python
student_id_card_generator('Swetha','Sanjay')
```

```python
# 8️⃣️ **Variable-Length Keyword Arguments → kwargs
# Used to receive named arguments as a dictionary.
def user_info(**details):
    print(details)

user_info(name="Nivi", age=25, city="Chennai")
```

```python
# 9️⃣ Functions Returning Multiple Values
def calculator(a, b):
    return a+b, a-b, a*b, a/b

add_, sub_, mul_, div_ = calculator(10, 5)
print(add_, sub_, mul_, div_)
```

```python
# 🔟 Lambda Functions (Anonymous Functions)
# Useful for short, one-line functions.
square = lambda num: num * num

print(square(4))
print(square(5))

# Another example with multiple arguments:
add = lambda a, b: a + b

print(add(10, 20))
```

```python
# 1️⃣1️⃣ Scope (Local vs Global Variables)
x = 10  # global

def test():
    y = 5  # local
    print(x, y)

test()
print(x) # will work because of gobal scope
# print(y) # will give error because it local scope variable inside test function it will work only inside test function
```

```python
# 1️⃣2️⃣ Functions Inside Functions
def outer():
    print("Outer function")

    def inner():
        print("Inner function")

    # inner() will execute only if call inside outer

# inner() # not defined because its local scope function inside outer function
outer()
```

```python
# 1️⃣3️⃣ Using global Keyword
count = 0

def increase():
    global count
    count += 1

increase()
print(count)
```

```python
increase()
print(count)
```

```python
# List Methods
fruits = ["apple", "banana"]
fruits.append("orange")
fruits.insert(0,"watermelon")
fruits.remove("banana")
print(fruits)

fruits.sort()
print(fruits)

fruits.reverse()
print(fruits)
```
