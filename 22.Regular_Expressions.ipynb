```python
import re

text = "Welcome to the Python course"

result = re.search("Python", text)
print(result)
print(result.span())
print(result.group())
```

```python
text = "apple, banana, apple, orange"

matches = re.findall("apple", text)
print(matches)
```

```python
# Character Patterns
# 🔹 Match digits
text = "A1B2C3abc@$%^#"
print(re.findall(r"\d", text))

# 🔹 Match alphabets
print(re.findall(r"[A-Z]", text))
print(re.findall(r"[a-z]", text))
print(re.findall(r"[A-Za-z]", text))

# 🔹 Match both
print(re.findall(r"[0-9]", text))
print(re.findall(r"[A-Za-z0-9]", text))
```

| Character   | Description       | Example    |
| ----------- | ----------------- | ---------- |
| \d          | Digit             | 123        |
| \D          | Non Digit         | ABC        |
| \w          | Alphanumeric      | user123    |
| \W          | Non Alphanumeric  | *+-=%$     |
| \s          | White Space       | z y z      |
| \S          | Non White Space   | dark       |

```python
contact_details = 'My phone number is +91 98756-12532'
pattern= r"\W\d\d\s\d\d\d\d\d\W\d\d\d\d\d"

re.search(pattern, contact_details)
```

# Quantifier
| Character   | Description            | 
| ----------- | ---------------------- |
| +           | One or More            |
| {3}         | Occurs exactly 3 Times |
| {2,4}       | Occurs 2 to 3 Times    |
| {3,}        | Occurs 3 or more Times | 
| *           | Occurs 0 or more Times |
| ?           | Once or None           | 

```python
pattern= r"\W{1}\d{2}\s\d{5}\W{1}\d{5}"

re.search(pattern, contact_details)
```

```python
# Validate Email
email = "test@gmail.com"

pattern = r"^[\w\.-]+@[\w\.-]+\.\w{2,}$"

if re.match(pattern, email):
    print("Valid Email")
else:
    print("Invalid Email")
```

```python
# Split using Regex
text = "one,two;three four"
result = re.split(r"[;, ]", text)

print(result)
```
