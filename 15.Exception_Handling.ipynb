```python
# Exceptions are runtime errors that stop the normal flow of a program.
print(10 / 0)   # ZeroDivisionError
```

```python
# 1️⃣ Basic try–except
try:
    print(10 / 0)
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

```python
# 2️⃣ Multiple except Blocks
# Used when you want to handle different errors differently.

try:
    num = int("Hello")
    print(10 / num)
except ValueError:
    print("Invalid number entered!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

```python
# 3️⃣ Catching ALL Errors (Not recommended for real apps, but useful for beginners)
try:
    print(x)
except Exception as e:
    print("Something went wrong:", e)
```

```python
# 4️⃣ try–except–else
# else runs only when no exception occurs.
try:
    num = int(input("Enter a number: "))
except ValueError:
    print("Invalid number!")
else:
    print("Square is:", num * num)
```

```python
# 5️⃣ try–finally
# finally runs always, no matter what.
# Useful for:
    # Closing a file
    # Closing DB connection
    # Cleaning resources

try:
    num = int(input("Enter a number: "))
    result = 100 / num
except ZeroDivisionError:
    print("You cannot divide by zero!")
except ValueError:
    print("Please enter a valid number!")
finally:
    print("Calculation attempt finished.")
```

```python
# 6️⃣ try + except + else + finally (All together)
try:
    num = int("50")
except ValueError:
    print("Invalid value")
else:
    print("Converted successfully:", num)
finally:
    print("Execution completed")
```

```python
# 7️⃣ raise – Manually Throwing Exceptions
# You can raise an exception yourself:

age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

```python
def withdraw(balance, amount):
    if amount > balance:
        raise Exception("Insufficient funds")
    return balance - amount

print(withdraw(1000, 500))
print(withdraw(1000, 2000))
```

```python
# 8️⃣ Creating Custom Exceptions
class InvalidAgeError(Exception):
    pass

def check_age(age):
    if age < 18:
        raise InvalidAgeError("User must be 18+")
    print("Valid age")

check_age(16)
```

```python
# 9️⃣ Real-World Examples
# ✔️ Login validation
def login(username, password):
    if username == "" or password == "":
        raise ValueError("Username or password missing!")

try:
    login("", "1234")
except ValueError as e:
    print("Login error:", e)
```
