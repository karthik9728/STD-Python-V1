```python
# 🟦 LIST (list)

# A list is an ordered, changeable (mutable) collection that allows duplicate items.
```

```python
fruits = ["apple", "banana", "orange"]
numbers = [10, 20, 30, 40]
mixed = [10, "hello", True, 99.5]

print(fruits)
print(numbers)
print(mixed)
```

```python
# 🔹 Accessing Items
print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
# print(fruits[-4]) Error: List index out of range
```

```python
# 🔹 List is Mutable
fruits[1] = "grapes"
print(fruits)
```

```python
# 🔹 Adding & Removing Items
# add at end
fruits.append("kiwi")     
print(fruits)
# add at index
fruits.insert(1, "watermelon")   
print(fruits)
# remove specific value
fruits.remove("apple")     
print(fruits)
# remove last item
fruits.pop()               
print(fruits)
# remove based on index
fruits.pop(1)               
print(fruits)
fruits.insert(1,"grapes")
print(fruits)
```

```python
# 🔹 Sorting & Reversing
fruits.sort()
print(fruits)
fruits.reverse()
print(fruits)

numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
```

```python
# 🔹 Find Length
print(f'The length of fruits list: {len(fruits)}')
print(f'The length of numbers list: {len(numbers)}')
print(f'The length of mixed list: {len(mixed)}')
```

```python
# 🔹 Slicing
print(fruits)
print(fruits[1:])
print(fruits[::-1])
print(numbers)
print(numbers[::-1])
```

```python
# list concatenation
shoppingList1=["bread","milk","cheese"]
shoppingList2=["playstation","xbox","iphone"]
shoppingList = shoppingList1 + shoppingList2
print(shoppingList)
```
