```python
import time

# Returns the number of seconds since Jan 1, 1970 (Unix epoch).
current_time = time.time()
print(current_time)
```

```python
# time.sleep(seconds) — Delay Execution
print("Start")
time.sleep(2)   # Pause program for 2 seconds
print("End after 2 seconds")
```

```python
# time.ctime() — Convert Timestamp to Readable Time

print(time.ctime())                 # current time
print(time.ctime(0))                # epoch time
print(time.ctime(time.time()))      # convert timestamp to readable
```

```python
# time.localtime() — Get Broken-Down Time Structure
t = time.localtime()
print(t)
print("Year:", t.tm_year)
print("Month:", t.tm_mon)
print("Day:", t.tm_mday)
print("Hour:", t.tm_hour)
```

```python
# time.strftime() — Format Time

t = time.localtime()

formatted = time.strftime("%Y-%m-%d %H:%M:%S", t)
print(formatted)
```

```python
print(time.strftime("%d-%m-%Y"))
print(time.strftime("%H:%M:%S"))
print(time.strftime("%A"))  # weekday
```

```python
start = time.time()

for i in range(5):
    print(i)
    time.sleep(1)

end = time.time()

print("Time taken:", end - start)
```
