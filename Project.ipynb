```python
import datetime

class Post:
    def __init__(self,post_id,title,content,created_on):
        self.post_id=post_id
        self.title=title
        self.content=content
        self.created_on=created_on

class NotFoundError(Exception):
    pass

class Blog:
    def __init__(self):
        self.__posts = []

    def view_post(self):
        if not self.__posts:
            print("No posts available.\n")
            return
        
        for  post in self.__posts:
            print(f"""
Post ID     : {post.post_id}
Title       : {post.title}
Content     : {post.content}
Created On  : {post.created_on}
----------------------------------------
""")

    def create_post(self):
        post_id = self.__posts[-1].post_id + 1 if self.__posts else 1
        title = input('Enter Post Title:\n')
        content = input('Enter Your Thoughts:\n')
        # created_on= datetime.datetime.now()
        created_on = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        new_post = Post(post_id,title,content,created_on)
        self.__posts.append(new_post)
        print('Post Created Successfully!!!')

    def edit_post(self):
        post_id = int(input('Enter Post Id\n'))
        post = next((p for p in self.__posts if p.post_id == post_id), None) 

        if(post):
            title = input('Enter Post Title:\n')
            content = input('Enter Your Thoughts:\n')
            post.title = title
            post.content = content
            print('Post Updated Successfully!!!')
        else:
            raise NotFoundError("Invalid Post Id\n")

    def delete_post(self):
        post_id = int(input('Enter the post id you want to delete\n'))
        post = next((p for p in self.__posts if p.post_id == post_id), None) 

        if(post):
            self.__posts.remove(post)
            print('Post Deleted Successfully!!!')
        else:
            raise NotFoundError("Invalid Post Id")
    
    def export_file(self):
        if not self.__posts:
            print("No posts to export.")
            return
            
#         f=open('posts.txt','w')
#         for post in self.__posts:
#             f.write(f"""
# Post ID     : {post.post_id}
# Title       : {post.title}
# Content     : {post.content}
# Created On  : {post.created_on}
# ----------------------------------------
# """)
#         f.close()
        print('File Export Successfully!!!')

    
    def start(self):
        while True:
            try:
                print('Select Your Option')
                print('1. Create Post')
                print('2. Edit Post')
                print('3. View Post')
                print('4. Delete Post')
                print('5. Export Post as Text File')
                print('6. Stop Application')
        
                selected_option = int(input())
        
                if selected_option == 1:
                    self.create_post()
                elif selected_option == 2:
                    self.edit_post()
                elif selected_option == 3:
                    self.view_post()
                elif selected_option == 4:
                    self.delete_post()
                elif selected_option == 5:
                    self.export_file()
                elif selected_option == 6:
                    print('Application Stopped!!')
                    break
                else:
                    print('Incorrect Option')
                    continue
            except ValueError as e:
                print(e)
                print('Restarting Application')
                continue
            except NotFoundError as e:
                print(e)
                print('Restarting Application')
                continue
        
```

```python
blog = Blog()
blog.start()
```
