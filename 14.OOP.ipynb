```python
class Person:
    # class attribute
    planet = "Earth"

    # All classes have a built-in method called __init__(), which is always executed when the class is being initiated.
    # The __init__() method is used to assign values to object attribute, or to perform operations that are necessary when the object is being created.

    # The self parameter is a reference to the current instance of the class. It is used to access attribute and methods that belong to the class.
    
    # def __init__(self,name,age,phone_no,location='Unknown',salary, company): # default paramter always comes at end
    def __init__(self,name,age,phone_no,salary,company,location="Unknown"):
        self.name = name  # instance attribute
        self.age = age
        self.phone_no = phone_no
        self.salary = salary
        self.company = company
        self.location = location

    # methods
    def get_details(self):
        print(f"{self.name} is {self.age} years old, from the {self.location} region, living on planet {self.planet}.")

    def get_employment_status(self):
        print(f"{self.name} is currently working at {self.company}.")

    def get_salary_details(self):
        print(f"{self.name}'s monthly salary is {self.salary}, and the annual salary is {self.salary * 12}.")
```

```python
p1 = Person("John",25,"9876543210",45000,"Google","USA")
p2 = Person("Jane",24,"9678543210",54000,"Meta")
```

```python
# Inheritance 
class Company(): # Base Class / Super Class / Parent Class

    def __init__(self,company_id,company_name,country):
        self.company_id=company_id
        self.company_name=company_name
        self.country=country

    def get_company_details(self):
        print(f'Company Detials\nId:{self.company_id}\nName:{self.company_name}\nLocation:{self.country}')
        print('--------------------------------')
```

```python
google = Company(1001,"Google","India")
google.get_company_details()

meta = Company(1002,"Meta","USA")
meta.get_company_details()
```

```python
# Polymorphism
class Vehicle():
    def start_engine(self):
        print('Engine is started!')

    def stop_engine(self):
        print('Engine is stopped!!!')

class Car(Vehicle):
    # pass

    # method overriding / run time polymorphism
    def start_engine(self):
        print('Car is started!')

    # method overriding / run time polymorphism
    def stop_engine(self,duration=10):
        print(f'Car engine will stops in {duration} seconds')

class Truck(Vehicle):
    # pass
    
    def start_engine(self):
        print('Truck is started!')

    def stop_engine(self):
        print('Truck is stopped!!!')

bmw = Car()
bmw.start_engine()
bmw.stop_engine(20)

volvo = Truck()
volvo.start_engine()
volvo.stop_engine()
```

```python
# Duck Typing — YES
# Python does not care about the type of object — if it has the method, it will work.
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for a in [Dog(), Cat()]:
    a.sound()
```

```python
# Operator Overloading — YES
# Python allows customizing operators (+, -, *, etc.) using magic methods.

class Box:
    def __init__(self, volume):
        self.volume = volume

    # operator overloading for +
    def __add__(self, other):
        return Box(self.volume + other.volume)

    # operator overloading for +
    def __sub__(self, other):
        return Box(self.volume - other.volume)
    
    # operator overloading for *
    def __mul__(self, other):
        return Box(self.volume * other.volume)

    # string representation of call
    def __str__(self):
        return f"Box Volume: {self.volume}"


b1 = Box(20)
b2 = Box(10)

result1 = b1 + b2   # calls __add__()
result2 = b1 - b2   # calls __sub__()
result3 = b1 * b2   # calls __mul__()

print(b1)
print(b2)
print(result1)
print(result2)
print(result3)
```

```python
# Encapsulation (Data Hiding)
# Use _ (protected) or __ (private).
class Bank:
    def __init__(self):
        self.__balance = 0   # private

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

b = Bank()
b.deposit(500)
print(b.get_balance())
# print(b.__balance)   ❌ error (private)
```

```python
class Account:
    def __init__(self):
        self._balance = 0      # protected attribute

    def deposit(self, amount):
        self._balance += amount

    def show_balance(self):
        print("Balance:", self._balance)

class SavingsAccount(Account):
    def apply_interest(self):
        self._balance += self._balance * 0.05
```

```python
acc = SavingsAccount()
acc.deposit(1000)

acc.apply_interest()
acc.show_balance()

print(acc._balance)

acc._balance = 50000

print(acc._balance)
```

```python
 # Abstraction means hiding the internal implementation details and showing only the essential features to the user.
 # 👉 User knows WHAT a method does
 # 👉 User does NOT care HOW it is implemented
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class GooglePay(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Google Pay")

class PhonePe(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using PhonePe")
```

```python
payment1 = GooglePay()
payment1.pay(500)

payment2 = PhonePe()
payment2.pay(1000)

# 📌 User only knows pay(amount)
# 📌 Implementation differs — abstraction achieved
```

| Concept           | Meaning                                 |
| ----------------- | --------------------------------------- |
| **Abstraction**   | Hide implementation, show functionality |
| **Encapsulation** | Hide data using access control          |

```python
# Normal Class with manual __eq__
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __eq__(self, other):
        return self.title == other.title and self.pages == other.pages


b1 = Book("Python Basics", 300)
b2 = Book("Python Basics", 300)

print(b1 == b2)
```

```python
class Laptop:

    def __init__(self, brand, ram, processor_model, processor_speed):
        self.brand = brand
        self.ram = ram
        # creating inner class object inside outer class
        self.processor = self.Processor(processor_model, processor_speed)

    def show_details(self):
        print(f"Laptop Brand: {self.brand}")
        print(f"RAM: {self.ram}GB")
        self.processor.show_processor()  # calling inner class method
        print("--------------------------------")

    # Inner Class
    class Processor:
        def __init__(self, model, speed):
            self.model = model
            self.speed = speed  # in GHz

        def show_processor(self):
            print(f"Processor Model: {self.model}")
            print(f"Processor Speed: {self.speed}GHz")


# Creating outer class object
l1 = Laptop("Dell", 16, "Intel i7", 3.5)

l1.show_details()
```
