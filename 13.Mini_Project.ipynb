```python
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 50:
        return "C"
    else:
        return "Fail"

name = input("Enter Student Name: ")
roll = input("Enter Roll Number: ")

subjects = ["English", "Maths", "Science", "History", "Computer"]
marks = []

for sub in subjects:
    score = int(input(f"Enter marks for {sub}: "))
    marks.append(score)

total = sum(marks)
avg = total / len(marks)
grade = calculate_grade(avg)

print("\n------ MARKSHEET ------")
print(f"Name       : {name}")
print(f"Roll No    : {roll}")
print(f"Total      : {total}")
print(f"Average    : {avg:.2f}")
print(f"Grade      : {grade}")
print("------------------------")
```

```python
balance = 0

def check_balance():
    print(f"Your Balance: {balance}")

def deposit():
    global balance
    amount = int(input("Enter amount to deposit: "))
    balance += amount
    print("Deposit Successful!")

def withdraw():
    global balance
    amount = int(input("Enter amount to withdraw: "))
    
    if amount > balance:
        print("Insufficient Balance!")
    else:
        balance -= amount
        print("Withdrawal Successful!")


while True:
    print("\n----- ATM MENU -----")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        check_balance()
    elif choice == "2":
        deposit()
    elif choice == "3":
        withdraw()
    elif choice == "4":
        print("Thank you for using the ATM!")
        break
    else:
        print("Invalid choice! Try again.")
```

```python
def process_order(name: str, price: float, quantity: int):
    print(f'Product: {name},  ₹{price} × {quantity} = ₹{price * quantity}')


orders = []   # list of tuples (name, price, qty)

while True:
    print("\n----- Select Product -----")
    print("1. Bread   - 50")
    print("2. Milk    - 60")
    print("3. Cheese  - 80")
    print("4. Check Cart")
    print("5. Checkout & Exit")

    choice = input("Choose an option: ")

    # Add Bread
    if choice == "1":
        quantity = int(input("Enter quantity: "))
        orders.append(("Bread", 50, quantity))
        print("Added Bread to cart!")

    # Add Milk
    elif choice == "2":
        quantity = int(input("Enter quantity: "))
        orders.append(("Milk", 60, quantity))
        print("Added Milk to cart!")

    # Add Cheese
    elif choice == "3":
        quantity = int(input("Enter quantity: "))
        orders.append(("Cheese", 80, quantity))
        print("Added Cheese to cart!")

    # Show cart
    elif choice == "4":
        print("\n----- CART -----")
        if len(orders) == 0:
            print("Cart is empty!")
        else:
            total = 0
            for item in orders:
                name, price, qty = item
                process_order(name, price, qty)
                total += price * qty
            print("-----------------")
            print(f"Total Amount: ₹{total}")

    # Checkout & Exit
    elif choice == "5":
        print("\n----- BILL -----")
        if len(orders) == 0:
            print("Cart is empty. Nothing to checkout.")
        else:
            grand_total = 0
            for item in orders:
                name, price, qty = item
                process_order(name, price, qty)
                grand_total += price * qty
            print("----------------------")
            print(f"Grand Total: ₹{grand_total}")
        print("Thank you! Goodbye.")
        break

    else:
        print("Invalid choice! Try again.")
```

```python
def c_to_f(celsius):
    return (celsius * 9/5) + 32

def f_to_c(fahrenheit):
    return (fahrenheit - 32) * 5/9

while True:
    print("\n--- Temperature Converter ---")
    print("1. Celsius to Fahrenheit")
    print("2. Fahrenheit to Celsius")
    print("3. Exit")
    
    choice = input("Choose an option: ")
    if choice == "1":
        c = float(input("Enter temperature in Celsius: "))
        print(f"{c}°C = {c_to_f(c):.2f}°F")
    elif choice == "2":
        f = float(input("Enter temperature in Fahrenheit: "))
        print(f"{f}°F = {f_to_c(f):.2f}°C")
    elif choice == "3":
        print("Thank you for using Temperature Converter!")
        break
    else:
        print("Invalid choice! Try again.")
```

```python
contacts = {}   # dictionary to store name : phone number


def add_contact():
    name = input("Enter Name: ")
    phone = input("Enter Phone Number: ")
    contacts[name] = phone
    print("Contact Added Successfully!")


def view_contacts():
    if len(contacts) == 0:
        print("No contacts available!")
    else:
        print("\n----- Contact List -----")
        for name, phone in contacts.items():
            print(f"{name} : {phone}")


def search_contact():
    name = input("Enter name to search: ")
    if name in contacts:
        print(f"{name}'s Number: {contacts[name]}")
    else:
        print("Contact not found!")


# Main Program
while True:
    print("\n----- CONTACT BOOK -----")
    print("1. Add Contact")
    print("2. View Contacts")
    print("3. Search Contact")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        add_contact()
    elif choice == "2":
        view_contacts()
    elif choice == "3":
        search_contact()
    elif choice == "4":
        print("Thank you for using Contact Book!")
        break
    else:
        print("Invalid option! Try again.")
```

```python
def process_order(name: str, price: float, quantity: int):
    """
    Process a single order line (prints the line).
    """
    line_total = price * quantity
    print(f'Product: {name}, ${price:.2f} * {quantity} = ${line_total:.2f}')


PRODUCTS = {
    "1": ("Bread", 50.0),
    "2": ("Milk", 60.0),
    "3": ("Cheese", 80.0),
    "4": ("Pepsi", 35.0),
    "5": ("Coke", 36.0),
    "6": ("Check Cart", None),
    "7": ("Exit", None),
}


def show_menu():
    print("\n----- Select Product ------")
    print(f"{'Option':<8}{'Product':<12}{'Price':>8}")
    for opt in ["1", "2", "3", "4", "5", "6", "7"]:
        name, price = PRODUCTS[opt]
        price_str = f"{price:.2f}" if price is not None else ""
        print(f"{opt:<8}{name:<12}{price_str:>8}")


def add_to_cart(cart: dict, product_key: str, qty: int):
    name, price = PRODUCTS[product_key]
    if product_key in cart:
        cart[product_key]["quantity"] += qty
    else:
        cart[product_key] = {"name": name, "price": price, "quantity": qty}
    print(f"Added {qty} x {name} to cart.")


def show_cart(cart: dict):
    if not cart:
        print("\nCart is empty.")
        return

    print("\n----- Your Cart -----")
    print(f"{'No.':<4}{'Product':<12}{'Qty':>6}{'Price':>10}{'Total':>12}")
    grand_total = 0.0
    for i, (k, item) in enumerate(cart.items(), start=1):
        name = item["name"]
        qty = item["quantity"]
        price = item["price"]
        total = price * qty
        grand_total += total
        print(f"{i:<4}{name:<12}{qty:>6}{price:>10.2f}{total:>12.2f}")
    print(f"{'':-^44}")
    print(f"{'Grand Total:':<32}${grand_total:.2f}")


def modify_cart(cart: dict):
    """
    Allow user to remove an item or change its quantity.
    """
    if not cart:
        print("Cart is empty.")
        return

    show_cart(cart)
    keys = list(cart.keys())
    while True:
        try:
            choice = input("\nEnter item number to modify (or press Enter to go back): ").strip()
            if choice == "":
                return
            idx = int(choice) - 1
            if idx < 0 or idx >= len(keys):
                print("Invalid item number.")
                continue
            key = keys[idx]
            item = cart[key]
            action = input(f"Modify '{item['name']}' — (r)emove, (u)pdate quantity, (c)ancel: ").strip().lower()
            if action == "r":
                del cart[key]
                print(f"Removed {item['name']} from cart.")
                return
            elif action == "u":
                new_q = input("Enter new quantity (0 to remove): ").strip()
                new_qi = int(new_q)
                if new_qi <= 0:
                    del cart[key]
                    print(f"Removed {item['name']} from cart.")
                else:
                    item["quantity"] = new_qi
                    print(f"Updated {item['name']} quantity to {new_qi}.")
                return
            else:
                print("Cancelled modification.")
                return
        except ValueError:
            print("Please enter a valid number.")


def checkout(cart: dict):
    if not cart:
        print("Cart is empty. Nothing to checkout.")
        return
    print("\n----- Checkout / Receipt -----")
    grand_total = 0.0
    for item in cart.values():
        process_order(item["name"], item["price"], item["quantity"])
        grand_total += item["price"] * item["quantity"]
    print(f"{'-'*36}")
    print(f"{'GRAND TOTAL':<24}${grand_total:.2f}")
    # After checkout we clear the cart
    cart.clear()
    print("Thank you for your purchase!")


def main():
    cart = {}
    while True:
        show_menu()
        choice = input("Choose an option: ").strip()

        if choice in {"1", "2", "3", "4", "5"}:
            # Add product
            try:
                qty_str = input("Please enter quantity: ").strip()
                qty = int(qty_str)
                if qty <= 0:
                    print("Quantity must be a positive integer.")
                    continue
            except ValueError:
                print("Invalid quantity. Enter a whole number.")
                continue

            add_to_cart(cart, choice, qty)

        elif choice == "6":
            # Check cart: show and allow modification or checkout
            while True:
                show_cart(cart)
                sub = input("\nOptions: (m)odify item, (c)heckout, (b)ack to menu: ").strip().lower()
                if sub == "m":
                    modify_cart(cart)
                elif sub == "c":
                    checkout(cart)
                    break
                elif sub == "b" or sub == "":
                    break
                else:
                    print("Unknown option. Please choose m, c or b.")

        elif choice == "7":
            confirm = input("Are you sure you want to exit? (y/n): ").strip().lower()
            if confirm == "y":
                # Offer to checkout if cart not empty
                if cart:
                    print("You have items in your cart.")
                    save = input("Would you like to checkout before exiting? (y/n): ").strip().lower()
                    if save == "y":
                        checkout(cart)
                print("Goodbye!")
                break
            else:
                continue
        else:
            print("Invalid choice! Try again.")


# if __name__ == "__main__":
#     main()

main()
```
