```python
# 1️⃣ What is a Generator?
# A generator is a special type of function that:
    # does not return all values at once,
    # produces values one at a time, only when needed (lazy evaluation),
    # uses yield instead of return.
# If we call next() after values are exhausted, Python raises: StopIteration

def generate_numbers():
    for x in range(1,3):
        yield x
```

```python
result=generate_numbers()
print(result)
print(next(result))
print(next(result))
print(next(result))
```

```python
# why we don't get error on for loop? because for loop automtically catch's this error and stop calling next
for num in generate_numbers():
    print(num)
```

Why Generators?
- ✔ Memory Efficient
- Does not store everything in RAM.
- Example: generate 1 million numbers.
- ❌ Normal List → Heavy
- ✅ Generator → Very Light

```python
# Infinite Generator
# You cannot create infinite lists, but you can create infinite generators:

def infinite_counter():
    num = 1
    while True:
        yield num
        num += 1

counter = infinite_counter()

print(next(counter))  
print(next(counter))  
print(next(counter))  
print(next(counter))  
```
