```python
from collections import defaultdict

colors = {'R':'RED','B':'Black'}
print(colors['R'])
# print(colors['G']) # Key Error

colors_default = defaultdict(lambda:'Unkown')
colors_default['R']='RED'
print(colors_default['R'])
print(colors_default['G'])
print(colors_default)
```

```python
# Tuple
# A normal tuple
person = ("John", 25, "India")

print(person[0])  # John
print(person[1])  # 25
print(person[2])  # India
```

❌ Problem with Tuple

You must remember the meaning of each index.

person[0] → name

person[1] → age

person[2] → country

If someone else reads your code, it's harder to understand.

```python
from collections import namedtuple

Person = namedtuple("Person", ["name", "age", "country"])

person = Person("John", 25, "India")

print(person.name)     # John
print(person.age)      # 25
print(person.country)  # India
```

✅ Benefits of namedtuple

More readable (person.name instead of person[0])

Still immutable like tuple

Lightweight (not heavy like class)

Fields are self-explanatory

```python
from collections import deque

queue = deque(["John", "Jane", "Mark"])
print(queue)
queue.append("Alex")   # Add right
print(queue)
queue.appendleft("Sara")  # Add left
print(queue)

queue.pop()         # Remove right
print(queue)
queue.popleft()     # Remove left
print(queue)
```

```python
from collections import ChainMap

defaults = {"theme": "light", "font": "Arial"}
user_settings = {"theme": "dark"}

config = ChainMap(user_settings, defaults)

print(config)

print(config["theme"])  # dark (overrides)
print(config["font"])   # Arial


# This means:
    # First search in user_settings
    # If not found, then search in defaults
    # If still not found → KeyError
    # So earlier dicts override later ones.
```

| Key   | user_settings | defaults | Result    | Why?                        |
| ----- | ------------- | -------- | --------- | --------------------------- |
| theme | dark          | light    | **dark**  | First dict overrides second |
| font  | (missing)     | Arial    | **Arial** | Falls back to next dict     |
