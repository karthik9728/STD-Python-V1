# A decorator is a function that adds extra features to another function without changing its original code.
Think of it like:

👉 Adding a wrapper around a gift

👉 Adding filters on an Instagram photo

👉 Adding toppings on a pizza


```python
def my_decorator(func):
    def wrapper():
        print("Before function runs...")
        func()
        print("After function runs...")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

✔ Explanation:

- my_decorator receives say_hello as func
- wrapper() adds extra behavior before and after calling func()
- @my_decorator automatically applies the wrapper
(same as writing: say_hello = my_decorator(say_hello))

```python
# Decorator with Arguments
def my_decorator(func):
    def wrapper(name):
        print("Preparing greeting...")
        func(name)
        print("Greeting complete.\n")
    return wrapper

@my_decorator
def greet(name):
    print(f"Hello {name}!")

greet("John")
```

```python
# Decorator for Multiple Arguments
def calc_decorator(func):
    def wrapper(a, b):
        print("Calculation started...")
        func(a, b)
        print("Calculation finished!")
    return wrapper

@calc_decorator
def add(a, b):
    print(a + b)

add(10, 20)
```
