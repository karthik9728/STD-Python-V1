1️⃣ Higher-Order Functions

A function that takes another function as argument or returns a function.

```python
def greet(name):
    return f"Hello, {name}"

def caller(func, value):
    return func(value)

print(caller(greet, "John"))
```

2️⃣ Pure vs Impure Functions

✔ Pure Function

NO side effects

Same input → Same output

```python
def pure_add(a, b):
    return a + b

print(pure_add(2, 3))  # Always 5
print(pure_add(2, 3))  # Always 5
print(pure_add(2, 3))  # Always 5
```

✖ Impure Function

Modifies something outside its scope

```python
count = 0

def impure_increment():
    global count
    count += 1
    return count

print(impure_increment())
print(impure_increment())
print(impure_increment())
```

3️⃣ Lambda, Map, Filter, Reduce

🔹 Lambda (Anonymous Function)

```python
square = lambda x: x*x
print(square(5))
```

🔹 map() → Apply function to each item

```python
numbers = [1, 2, 3, 4]
result = list(map(lambda x: x*2, numbers))
print(result)
```

🔹 filter() → Keep items that return True

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x%2 == 0, numbers))
print(evens)
```

🔹 reduce() (needs functools)

```python
from functools import reduce

result = reduce(lambda a,b: a+b, [1,2,3,4])
print(result)
```

4️⃣ Closure Functions

A function inside another function that remembers the outer variable even after outer function ends.

```python
def outer(msg):
    def inner():
        print("Message:", msg)
    return inner

greet = outer("Hello Python")
greet()  # remembers "Hello Python"
```

5️⃣ Partial Functions

Used to fix some arguments of a function in advance.

```python
from functools import partial

def power(base, exp):
    return base ** exp

square = partial(power, exp=2)
cube = partial(power, exp=3)

print(square(5))
print(cube(5))
```

6️⃣ Function Composition

Calling one function inside another to create a pipeline of logic.

```python
def double(x):
    return x * 2

def add_five(x):
    return x + 5

def compose(f, g):
    return lambda x: f(g(x))

new_func = compose(double, add_five)
print(new_func(10))  # (10 + 5) * 2 = 30
```

7️⃣ Callback Functions

A function passed into another function and called later.

```python
def download(callback):
    print("Downloading...")
    callback()

def notify():
    print("Download Completed!")

download(notify)
```

8️⃣ Recursive Functions

A function that calls itself.

Example: Factorial

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```
