```python
# A Boolean (or bool) is a datatype that can hold one of two values: True or False.
# Used to represent binary states, yes/no answers, on/off switches, or conditions.

# Real-World Example
# 1) A lamp is either ON or OFF.
# 2) A user’s login attempt is either successful (True) or failed (False).
```

```python
# Note: True and False are Python keywords (capitalized).
is_active = True
is_admin = False

print(is_active)
print(is_admin)
```

```python
x=(10>3)
print(x)
```

```python
y=bool(0) # False (0 is falsy) # Ture (1 is truthy)
print(y)
```

```python
z = bool("hello")   # True (non-empty string)
# z = bool("")
print(z)

# None is special constant in python
print(bool(None))
print(type(None))
```

```python
# How Booleans Are Created
# 1)Direct assignment (x = True)
# 2)Comparison operators (5 > 3 → True)
# 3)Logical expressions ((a > b) and (c == d))
# 4)Built-in functions (bool(value))
```
