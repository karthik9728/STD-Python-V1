A thread is a smaller unit of a program that can run “at the same time” as other parts.

In Python, threads are best for I/O-bound tasks:

Waiting for API response

Reading/Writing files

Sleeping / delays

CPU-heavy tasks don’t get big benefit from threads because of the GIL (Global Interpreter Lock), but no need to go very deep into theory.

# 1) Normal (Single Thread) vs Multi-thread

```python
import time

def print_numbers():
    for i in range(1, 6):
        print(f"[numbers] {i}")
        time.sleep(1)

def print_letters():
    for ch in ['A', 'B', 'C', 'D', 'E']:
        print(f"[letters] {ch}")
        time.sleep(1)


start = time.time()

print_numbers()
print_letters()

end = time.time()
print("Single-thread total time:", end - start)
```

# Now same work with threads:

```python
import time
import threading

def print_numbers():
    for i in range(1, 6):
        print(f"[numbers] {i}")
        time.sleep(1)

def print_letters():
    for ch in ['A', 'B', 'C', 'D', 'E']:
        print(f"[letters] {ch}")
        time.sleep(1)

start = time.time()

t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letters)

t1.start()   # start both at same time
t2.start()

t1.join()    # wait for both to finish
t2.join()

end = time.time()
print("Multi-thread total time:", end - start)
```
