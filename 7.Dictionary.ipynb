```python
# 🔶 Dictionary (dict)

# A Dictionary in Python is a collection of data stored as key–value pairs.
# It is used to store structured information where each value can be accessed using a unique key.
```

```python
colors={'r':'red','g':'green','b':'blue','p':'pink'}
# access direct based on key
print(colors)
print(colors['r'])
print(colors['g'])
print(colors['b'])

# Direct access
# print(colors['s']) # Key Error

# Safe access using get()
print(colors.get('s'))       # None (safe)
print(colors.get('s', 'Key not found'))  # with default message
```

```python
# ✅ 1. Add new item
colors['y'] = 'yellow'
print(colors)

# ✅ 2. Update existing item
colors['b'] = 'sky blue'
print(colors)

colors.update({'p': 'purple'})
print(colors)

# ✅ 3. Remove / Delete item
del colors['r']
print(colors)

# ❌ Using pop (also returns removed value)
colors.pop('g')
print(colors)
```

```python
# find avaiable keys in dictonary
print(colors.keys())

# find avaiable values in dictonary
print(colors.values())

# dictonary items
print(colors.items())
```

```python
employees={"John":{"Department":"IT","Salary":50000},"Jane":{"Department":"HR","Salary":45000}}
print(employees)
print(employees["John"])
print(employees["Jane"])
```

```python
# accessing nested value
print(employees["John"]["Department"])
print(employees["John"]["Salary"])
```

```python
movies = {
    "Avengers": {
        "Rating": 8.5,
        "TicketPrice": 250,
        "Category": ["Action", "Sci-Fi", "Adventure"],
        "Language": "English"
    },
    "Leo": {
        "Rating": 7.9,
        "TicketPrice": 200,
        "Category": ["Thriller", "Action"],
        "Language": "Tamil"
    },
    "KGF": {
        "Rating": 8.2,
        "TicketPrice": 220,
        "Category": ["Action", "Drama"],
        "Language": "Kannada"
    }
}

print(movies)
```

```python
print(f'Avengers movie details: {movies["Avengers"]}')
```

```python
print(f'Avengers movie rating: {movies["Avengers"]["Rating"]}')
```

```python
print(f'Avengers movie category: {movies["Avengers"]["Category"]}')
print(f'Avengers movie 1st category: {movies["Avengers"]["Category"][0]}')
```
