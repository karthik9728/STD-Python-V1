```python
# 1️⃣ OUTPUT using print()

# ✔ Basic Output
print("Hello, Python!")

# ✔ Printing multiple values
print("Name:", "John", "Age:", 25)

# ✔ Using variables
name = "Alice"
age = 22

print(name, age)
```

```python
# 2️⃣ INPUT using input()
name = input("Enter your name: ")
print("Welcome", name)
```

```python
# 3️⃣ Type Conversion with input()

# ✔ Integer Input
age = int(input("Enter your age: "))
print("Next year your age will be:", age + 1)
```

```python
# ✔ Float Input
price = float(input("Enter price: "))
print("Price with tax:", price + 5.0)
```

```python
# 4️⃣ Multiple Inputs in One Line
a, b = input("Enter two numbers: ").split()
print(a, b)
print(type(a),type(b))
```

```python
# ✔ Convert them into int
a, b = map(int, input("Enter two numbers: ").split())
print(a + b)
print(type(a),type(b))
```

```python
# 5️⃣ Formatting Output
name = "John"
city = "Chennai"

print(f"{name} is from {city}.")
```

```python
# 6️⃣ print() with end sep
# ✔ end parameter
# Used to avoid newline.

for i in range(5):
    print(i, end=" ")
```

```python
# ✔ sep parameter
# Used to change the separator between printed values.
print("apple", "banana", "cherry", sep=" / ")
```

```python
# 7️⃣ Escape Characters
print("Hello\nWorld")     # new line
print("Hello\tWorld")     # tab space
print("He said \"Python is fun\"")  # quotes
print('He said "Python is fun"')  # quotes
print("C:\\Users\\karthik")   # backslash
```

```python
# 8️⃣ Output Formatting (Advanced)
# ✔ width & alignment
# Useful for tables and clean output.

print(f"{'Name':<10}{'Score':>5}")
print(f"{'Arun':<10}{95:>5}")
print(f"{'Nivi':<10}{88:>5}")
```

```python
# ✔ Decimal formatting
pi = 3.1415926
print(f"{pi:.2f}")
print(f"{pi:10.3f}")
```
