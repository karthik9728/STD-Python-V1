```python
# Accessing Characters (Indexing)
name="Nivi"
print(name)
print(name[0])
print(name[3])
# string index out of range
# print(name[4])

# Strings Are Immutable
# Once created, you cannot change a string directly.
# You can only create a new string.
name="Karthik"
print(name)
# 'str' object does not support item assignment ❌ Not allowed
# name[0]="M"
```

```python
# string concatenation
firstName="Arun"
lastName="Kumar"
userName = firstName +" "+ lastName
print(userName)

print(3 + 5)
print('3' + '5')
# print(3 + '5') # unsupported operand type(s) for +: 'int' and 'str'
```

```python
# Slicing Strings
website="ByteLyn"
print(website[0:4]) # Byte
print(website[4:7]) # Lyn

alphabets="abcdefghijklmnopqrstuvwxyz"
print(len(alphabets))
print(alphabets[::]) # starting to end
print(alphabets[::2]) # 3rd parameter as jump size
print(alphabets[::3]) # 3rd parameter as jump size
print(alphabets[5:20:2]) 
print(alphabets[::-1]) # reverse string
```

```python
# Find Substring
text2="python course"
print(text2.find("python"))
print(text2.find("course"))
print(text2.find("dotnet")) # -1 for if not found

# String Membership Operators
print("python" in text)     # True
print("Python" in text)     # False (case sensitive)
print("java" not in text)   # True

# capitalize(), title()
print(text2.capitalize())
print(text2.title())
```

```python
# startswith() / endswith()
text3="You are your only limit."
print(text3.startswith("Yo")) # True
print(text3.startswith("yo")) # False (case sensitive)
print(text3.endswith("it"))   # False
print(text3.endswith("it."))  # True
```

```python
# string interpoltion
# 1) .format() method
color="orange"
print("Every aircraft must have a black box, and its color will be {}.".format(color))

model = "Airbus A320"
capacity = 180
print("The {} can carry up to {} passengers.".format(model, capacity))

print("The {} laptop comes with {} RAM and {} of storage.".format('Lenovo', '16GB', '512GB'))

print("The colors I've chosen are {0}, {1}, and {2}.".format('blue', 'red', 'purple'))
print("The colors I've chosen are {1}, {2}, and {0}.".format('blue', 'red', 'purple'))
print("The colors I've chosen are {2}, {0}, and {1}.".format('blue', 'red', 'purple'))
print("The colors I've chosen are {b}, {r}, and {p}.".format(b='blue', r='red',p='purple')) # with keyword
print("The colors I've chosen are {p}, {p}, and {p}.".format(b='blue', r='red',p='purple')) # with keyword

pi=3.14159265359
# Notice the space between = and the number — because print() inserts a space automatically.
print("The value of PI=",pi)
# Here, Python formats the text and places pi exactly where {} is.
print("The value of PI={}".format(pi))
print("The value of PI={:1.3f}".format(pi))
print("The value of PI={:1.2f}".format(pi))
print("The value of PI={:10.2f}".format(pi)) # here 10 is width
```

```python
# 2) f-string (formatted string literals)
print(f"Every aircraft must have a black box, and its color will be {color}.")
print(f"The {model} can carry up to {capacity} passengers.")
print(f"The {'Lenovo'} laptop comes with {'16GB'} RAM and {'512GB'} of storage.")
print(f"The value of PI={pi}")
print(f"The value of PI={pi:1.2f}")
```
