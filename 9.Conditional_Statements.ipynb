```python
# 1️⃣ Basic if Statement
temperature = 35
if temperature > 30:
    print("It's a hot day!")
```

```python
# 2️⃣ if-else

# Odd or Even
number = 10
if(10 % 2 == 0):
    print(f'{number} is even number')
else:
    print(f'{number} is odd number')
```

```python
order_amount = 450
if order_amount >= 500:
    print("You get FREE delivery!")
else:
    print("Delivery charge ₹49 will be added.")
```

```python
withdraw = 2000
balance = 1500
if withdraw <= balance:
    print("Withdrawal successful!")
else:
    print("Insufficient balance!")
```

```python
battery = 18
if battery < 20:
    print("Battery Low! Please plug in charger.")
else:
    print("Battery level is good.")
```

```python
# 3️⃣ if-elif-else
time = 19

if time < 12: 
  print('Good morning!')
elif time < 18: 
  print('Good afternoon!')
else: 
  print('Good evening!')


speed = 85   # km/h

if speed > 120:
    print("Over Speeding – Heavy Fine!")
elif speed > 80:
    print("Warning – Reduce Speed.")
elif speed > 40:
    print("Normal Speed.")
else:
    print("Too Slow – Speed Up!")
```

```python
# 4️⃣ Multiple Conditions using Logical Operators
age = 22
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Permission denied")
```

```python
# 5️⃣ Nested if
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to vote")
    else:
        print("Not a citizen")
```

```python
# Decision logic
age = 21
is_with_parent = False;
has_id_proof = False;

if age >= 18: 
    if has_id_proof: 
        print('You can visit the mall and can able to watch the movie');
    else: 
        print('You can visit the mall');
else: 
    if is_with_parent: 
        print('You can visit the play area.');
    else: 
        print('You are not allowed in the play area.');
```

```python
# 6️⃣ Using membership operator inside condition
fruits = ["apple","banana","orange"]

if "apple" in fruits:
    print("Apple available")
```

```python
# 🔷 match / case (Structural Pattern Matching)

command = "start"

match command:
    case "start":
        print("Starting...")
    case "stop":
        print("Stopping...")
    case _:
        print("Unknown command")
```

```python
drink_choice = "tea"

if drink_choice == "coffee":
    print("Order taken for coffee.")
elif drink_choice == "tea":
    print("Order taken for tea.")
elif drink_choice == "water":
    print("Order taken for water.")
else:
    print("Choose a drink")

match drink_choice:
    case "coffee":
        print("Order taken for coffee.")
    case "tea":
        print("Order taken for tea.")
    case "water":
        print("Order taken for water.")
    case _:
        print("Choose a drink")
```

```python
# 🔹 Literal & OR patterns
size = "M"

match size:
    case "S" | "Small":
        print("You selected a Small pizza.")
    case "M" | "Medium":
        print("You selected a Medium pizza.")
    case "L" | "Large":
        print("You selected a Large pizza.")
    case _:
        print("Invalid size selected.")
```

```python
day = "Sunday"

match day:
    case "Saturday" | "Sunday":
        print("It's a weekend!")
    case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
        print("It's a weekday.")
    case _:
        print("Not a valid day")
```

```python
vehicle = "bike"

match vehicle:
    case "car" | "jeep" | "van":
        print("Four-wheeler selected.")
    case "bike" | "scooter":
        print("Two-wheeler selected.")
    case _:
        print("Unknown vehicle type.")
```
