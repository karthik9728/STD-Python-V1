```python
# 🟧 TUPLE (tuple)

# A tuple is an ordered, immutable collection.
# Once created, you cannot modify its elements.
# tuple use () parenthesis
```

```python
colors=('red','blue','green')
numbers=(10,20,30,40,50,20,70,100,20)
print(colors)
print(type(colors))
```

```python
# 🔹 Tuples Are Immutable
# colors[0] = "yellow"   # ❌ Error
```

```python
mixed=("Toys",3.1415,255)
print(mixed)
```

```python
# 🔹 Accessing Items
print(colors[0])
print(colors[1])
print(colors[2])
# print(colors[3]) # Error tuple index out of range
```

```python
# 🔹 Sorting 
# colors.sort()  # ❌ Error'tuple' object has no attribute 'sort'
# print(colors)
```

```python
# 🔹 Tuple Operations
print(f'The legth of colors tuple: {len(colors)}')
print(colors.count("red"))
print(colors.index("green"))
print(f'The legth of numbers tuple:{len(numbers)}')
print(numbers.count(20))
print(numbers.index(20)) # index() always returns the index of the first occurrence.
```

```python
# 🔹 Slicing
print(numbers)
print(numbers[3:])
print(numbers[3:6])
print(numbers[::-1])
```

```python
# 🔹 Packing & Unpacking
person=("Roshan",26,1997)
name, age, dob = person 
# name, age , dob, location = person # ❌ Error not enough values to unpack
print(name)
print(age)
print(dob)
```

```python
# tuple concatenation
# its valid, because its not modify tuple it creates as new tuple 
shopping_list_one=("bread","milk","cheese")
shopping_list_two=("playstation","xbox","iphone")
shopping_list = shopping_list_one + shopping_list_two
print(shopping_list)

# its valid, because its not modify tuple it creates as new tuple 
container=("tv","printer","moniter")
container+=("mouse","keyboard","speakers")
print(container)

# 🧠 Why tuple works even though it’s immutable?
# Because:
    # You are NOT changing the internal elements
    # You are only reassigning the variable to a new tuple
# Immutability means the values inside cannot change.
# But assigning a new object to the variable is allowed.
```
