Threads share the same memory inside one process.

Processes are separate Python interpreters:

Each has its own memory space.

Can truly run in parallel on multiple CPU cores.

Multiprocessing is good for CPU-heavy tasks:

Calculations, simulations, image processing, etc.

```python
from multiprocessing import cpu_count

print("CPU cores available:", cpu_count())
```

```python
from multiprocessing import Process
import os
import time

def show_info(name):
    print(f"Task '{name}' is running in process id: {os.getpid()}")
    time.sleep(2)
    print(f"Task '{name}' finished.")

if __name__ == "__main__":
    # On Windows, this guard is REQUIRED
    p1 = Process(target=show_info, args=("Job-1",))
    p2 = Process(target=show_info, args=("Job-2",))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("Main process done.")
```
