# File handling allows Python programs to create, read, write, and modify files.
| Mode   | Meaning       | Use                                    |
| ------ | ------------- | -------------------------------------- |
| `"r"`  | Read          | Opens file for reading (default)       |
| `"w"`  | Write         | Creates new file / overwrites existing |
| `"a"`  | Append        | Adds new content at end                |
| `"x"`  | Create        | Creates file, errors if exists         |
| `"r+"` | Read + Write  | No truncation                          |
| `"w+"` | Write + Read  | Truncates                              |
| `"a+"` | Append + Read | Pointer at end                         |


```python
# 1️⃣ open() + read() → Reading a File
f=open('example.txt','r') 
content=f.read()
print(content)
f.close()
```

```python
# Reading Part of a File
f=open('example.txt','r') 
content=f.read(20)  # reads first 20 characters , space also consider as characters
print(content)
f.close()
```

```python
# 🔹 read() + readline() + readlines()
f=open('example.txt','r') 
print(f.readline())      # Reads first line
print(f.readline())      # Reads second line
print(f.readline())      # Reads third line
f.close()
```

```python
f = open("example.txt", "r")
lines = f.readlines()     # returns list of lines
print(lines)
f.close()
```

```python
# 2️⃣ write() → Writing to a File
```

```python
f=open('notes.txt','w')
f.write('Hello, this is a new file.\n')
f.write('We are learning Python file handling.')
f.close()
```

```python
# 3️⃣ append() → Adding Content Without Deleting Existing Data
f = open("notes.txt", "a")
f.write("\nThis line was appended later.")
f.close()
```

4️⃣ with open() → Context Manager (Best Practice)

✔ Automatically closes the file
✔ Cleaner code
✔ Prevents leaks/errors

```python
with open("notes.txt", "r") as f:
    print(f.read())
```

```python
try:
    with open("missing.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("The file does not exist!")
```

```python
filename = "student_notes.txt"

while True:
    print("\n1. Write Note")
    print("2. View Notes")
    print("3. Exit")

    choice = input("Choose: ")

    if choice == "1":
        note = input("Enter note: ")
        with open(filename, "a") as f:
            f.write(note + "\n")
        print("Note added!")

    elif choice == "2":
        try:
            with open(filename, "r") as f:
                print("\n--- Notes ---")
                print(f.read())
        except FileNotFoundError:
            print("No notes found!")
            
    elif choice == "3":
        break
    else:
        print("Invalid choice")
```
