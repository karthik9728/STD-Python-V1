```python
# 🟩 SET (set)

# A set is an unordered, unique collection of items.
# Duplicates are removed automatically.
```

```python
numbers=set()
```

```python
numbers.add(10)
print(numbers)
```

```python
numbers.add(20)
print(numbers)
```

```python
numbers.add(20) # Duplicates are removed automatically.
print(numbers)
```

```python
numbers.remove(10)
print(numbers)
```

```python
numbers.add(True)
print(numbers)
```

```python
numbers.add(1)  # Duplicates not allowed because 1 is True
print(numbers)
```

```python
numbers.remove(1) 
print(numbers)
```

```python
numbers.add(0)
print(numbers)
```

```python
numbers.add(False)  # Duplicates not allowed because 0 is False
print(numbers)
```

```python
# Set does not support indexing
s = {10, 20, 30}
# print(s[0])   # ❌ Error: 'set' object is not subscriptable
```

```python
# 🔹 Set Operations (Very Important)
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))         # {1,2,3,4,5}
print(a.intersection(b))  # {3}
print(a.difference(b))    # {1,2}
print(b.difference(a))    # {4,5}
```

```python
num_set= set()
print(num_set)
```

```python
# num_set.add([1,2,3])  # ❌ Error: unhashable type: 'list'
# Reason:
# Lists are mutable, so Python cannot safely calculate a stable hash value for them.
# Sets only allow immutable, hashable items.

num_set.add((1,2,3)) # 'tuple' can be added to a set
# Reason:
# Tuple is immutable → hashable → safe for set.
```

```python
print(num_set)
```

```python
num_set.add((4,5,6))
num_set.add((7,8,9))
print(num_set)
```

```python
num_set.add((1,2,3)) # duplicate not allowed
print(num_set)
```

```python
num_set.add((3,2,1)) # ✔ allowed, not duplicate
print(num_set)
```

```python
num_set.add(1)
print(num_set)
```
